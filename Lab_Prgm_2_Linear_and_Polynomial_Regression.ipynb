**Python Code WITHOUT Machine Learning **

```python
# Linear Regression from scratch

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]

n = len(x)

# Calculate means
x_mean = sum(x) / n
y_mean = sum(y) / n

# Calculate slope
numerator = 0
denominator = 0

for i in range(n):
    numerator += (x[i] - x_mean) * (y[i] - y_mean)
    denominator += (x[i] - x_mean) ** 2

b1 = numerator / denominator

# Calculate intercept
b0 = y_mean - b1 * x_mean

print("Slope:", b1)
print("Intercept:", b0)

# Prediction
predictions = []

for value in x:
    prediction = b0 + b1 * value
    predictions.append(prediction)

print("Predictions:", predictions)

# Calculate MAE
mae = 0

for i in range(n):
    mae += abs(y[i] - predictions[i])

mae = mae / n

# Calculate MSE
mse = 0

for i in range(n):
    mse += (y[i] - predictions[i]) ** 2

mse = mse / n

# RMSE
rmse = mse ** 0.5

# R2
ss_res = 0
ss_tot = 0

for i in range(n):
    ss_res += (y[i] - predictions[i]) ** 2
    ss_tot += (y[i] - y_mean) ** 2

r2 = 1 - (ss_res / ss_tot)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

**Linear Regression Using Built-in Scikit-learn**

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

# Data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 5, 4, 5])

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predictions
y_pred = model.predict(X)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)

print("Predictions:", y_pred)

# Evaluation
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

**Polynomial Regression Using Scikit-learn**

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1, 4, 9, 16, 25])

# Convert X into polynomial features
poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

print("Polynomial Features:")
print(X_poly)

# Create Linear Regression model
model = LinearRegression()

# Train model
model.fit(X_poly, y)

# Prediction
y_pred = model.predict(X_poly)

print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)

# Evaluation
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```
